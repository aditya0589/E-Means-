# Statistical Rigor & Hypothesis TestingAnalyzing 10 repeated independent trials across distinct random seeds:- Descriptive distributions (mean, standard deviation, 95% confidence intervals)- Paired Student's t-test and Wilcoxon signed-rank test for significance

```python
import sys
from pathlib import Path
sys.path.insert(0, "..")
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
agg_file = Path("results/processed/aggregated_results.csv") if Path("results/processed/aggregated_results.csv").exists() else Path("../results/processed/aggregated_results.csv")
df_agg = pd.read_csv(agg_file)
df_agg
```

```python
stat_file = Path("results/processed/statistical_results.csv") if Path("results/processed/statistical_results.csv").exists() else Path("../results/processed/statistical_results.csv")
df_stat = pd.read_csv(stat_file)
df_stat
```

```python
runs_file = Path("results/raw/repeated_runs.csv") if Path("results/raw/repeated_runs.csv").exists() else Path("../results/raw/repeated_runs.csv")
df_runs = pd.read_csv(runs_file)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5), dpi=150)
sns.boxplot(data=df_runs[df_runs["dataset"] == "wine"], x="algorithm", y="inertia", hue="algorithm", legend=False, palette="Set2", ax=ax1)
ax1.set_title("Wine: Inertia Distribution (10 Seeds)", fontsize=10.5, fontweight="bold")
ax1.set_xticklabels(ax1.get_xticklabels(), rotation=15, ha="right", fontsize=8.5)
ax1.grid(axis="y", linestyle="--", alpha=0.5)

sns.boxplot(data=df_runs[df_runs["dataset"] == "digits"], x="algorithm", y="total_time_sec", hue="algorithm", legend=False, palette="Set3", ax=ax2)
ax2.set_title("Digits: Runtime Distribution (10 Seeds)", fontsize=10.5, fontweight="bold")
ax2.set_xticklabels(ax2.get_xticklabels(), rotation=15, ha="right", fontsize=8.5)
ax2.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
