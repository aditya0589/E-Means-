## Findings & Discussion### 1. Epoch Efficiency- **E-Means completes training in exactly 5 epochs** via automatic scaling laws, compared to 15–45 iterations required by Lloyd's algorithm.- The 1-pass exact finalization step clears stochastic mini-batch noise, producing final inertias within 0.1% to 1.0% of the full-batch Lloyd solution.### 2. Computational Pruning vs Algorithmic Updates- **Elkan's algorithm** maintains exact parity with K-Means++ while eliminating **60% to 94.5%** of explicit distance calculations using triangle inequality bounds.- Bound maintenance incurs small array indexing overhead in low dimensions ($d \le 4$), but provides substantial gains in higher dimensions ($d \ge 25$).### 3. Stability Across Seeding- K-Means++ probabilistic seeding and E-Means reduce inertia variance across independent seeds by **> 3.5×** relative to Lloyd's uniform random sampling.### 4. Scope & Context- On CPU architectures, E-Means demonstrates linear scaling and small memory footprint.- On massive embedding datasets ($n \ge 10^7, d \ge 128$), as shown by Bessa et al.
